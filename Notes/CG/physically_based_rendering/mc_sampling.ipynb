```python
import numpy as np
import seaborn as sns
rng = np.random.default_rng()
```

```python
# semi circle of radius 2, 2pi
def f(x):
   return np.sqrt(4 - np.power(x,2))

def plot(data):
    sns.pointplot(
        data=data, 
        x="x", 
        y="y", 
        join=False,    # Removes the line connecting the dots
       # marker="o",          # Ensures the mean is a dot
        errorbar="sd",       # Can be "sd", "se", or "ci"
        capsize=0.1          # Adds horizontal caps to the error bars
    )
```

```python
# plot of the function
x = np.linspace(-2, 2, num=50)
sns.lineplot(x = x, y=f(x))
```

$$\frac{1}{n} \sum_{i = 1}^{N} \frac{f(x_i)}{p(x_i)}$$

```python
def mc_uniform(sample):
    return np.sum(f(rng.uniform(size=sample, low=-2, high=2.0))) * 4 / sample

mc_uniform_data = {'x': [], 'y': []}
for sample in range(10, 50):
    for _ in range(10):
        mc_uniform_data['x'].append(sample)
        mc_uniform_data['y'].append(mc_uniform(sample))
```

```python
plot(mc_uniform_data)
```
